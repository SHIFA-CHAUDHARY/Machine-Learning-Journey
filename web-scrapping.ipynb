```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers= {"User-Agent": "Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWeKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36"}
web_page = requests.get("https://www.ambitionbox.com/list-of-companies?",headers=headers).text
```

```python
soup = BeautifulSoup(web_page, 'html.parser')
```

```python
company=soup.find_all('div',class_='comapny-content-wrapper')
```

```python
len(company)
```

```python
name=[]
rating=[]
review=[]
ctype=[]
old=[]
hq=[]
employee=[]
for i in company:
    name.append(i.find('h2')).text.strip()
    rating.append(i.find('p',class_='rating').text.strip())   
    review.append(i.find('a'),class_='review-count').text.strip()
    ctype.append(i.find_all('p',class_='infoEntity')[0].text.strip())
    old.append(i.find_all('p',class_='infoEntity')[1].text.strip())
    hq.append(i.find_all('p',class_='infoEntity')[2].text.strip())
    employee.append(i.find_all('p',class_='infoEntity')[3].text.strip())
```

```python
company[0].find_all('p',class_='infoEntity')
```

```python
d={'name':name,'rating':rating,'review':review,'ctype':ctype,'old':old,'hq':hq,'employee':employee}
df = pd.DataFrame(d)
df 
```
